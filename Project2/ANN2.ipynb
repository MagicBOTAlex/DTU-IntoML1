```python
import pandas as pd
import torch
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import KFold
from tqdm import tqdm
import pandas as pd
from IPython.display import display
from sklearn.model_selection import LeaveOneOut
from sklearn.model_selection import KFold
from sklearn.preprocessing import MinMaxScaler
import torch
from torch.utils.data import DataLoader, TensorDataset
from tqdm import tqdm
import matplotlib.pyplot as plt
import numpy as np
```

```python
def map_value(x, in_min, in_max, out_min, out_max):
    return (x - in_min) * (out_max - out_min) / (in_max - in_min) + out_min
```

```python
import pandas as pd
import torch
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import KFold
from tqdm import tqdm

# -----------------------------
# 1. Load CSV and preprocess data
# -----------------------------
df = pd.read_csv('null-corrected.csv', index_col=0)
X = df.iloc[:, :-1]
Y = df.iloc[:, -1:]

# Normalize features and targets to [-1, 1]
scaler_X = MinMaxScaler(feature_range=(-1, 1))
X_scaled = scaler_X.fit_transform(X)

scaler_Y = MinMaxScaler(feature_range=(-1, 1))
Y_scaled = scaler_Y.fit_transform(Y)

# Set device to CUDA if available, otherwise CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# Convert features and targets to tensors
X_tensor = torch.tensor(X_scaled, dtype=torch.float32).to(device)
Y_tensor = torch.tensor(Y_scaled, dtype=torch.float32).to(device)

# -----------------------------
# 2. Set up outer CV (10 folds) and candidate hyperparameters
# -----------------------------
outer_cv = KFold(n_splits=10, shuffle=True, random_state=42)
folds = list(outer_cv.split(X_scaled))
print("Number of outer folds:", len(folds))

# Define candidate hyperparameters for inner CV tuning
candidate_hidden_units = [128, 256, 512, 1024, 1024*2]
candidate_regularizations = [0.005, 0.1]
learning_rate = 0.1

# Set number of epochs for inner and outer training
inner_epochs = 25   # fewer epochs to save time during hyperparameter tuning
outer_epochs = 50  # final training epochs on the full outer training set

# List to store results for each outer fold
results = []

# -----------------------------
# 3. Outer CV loop
# -----------------------------
for fold_idx, (train_index, test_index) in enumerate(tqdm(folds, desc="Outer CV Folds"), start=1):
    # Split data into outer training and test sets
    X_train = X_tensor[train_index]
    Y_train = Y_tensor[train_index]
    X_test = X_tensor[test_index]
    Y_test = Y_tensor[test_index]
    
    # -----------------------------
    # 3a. Inner CV for hyperparameter tuning
    # -----------------------------
    inner_cv = KFold(n_splits=3, shuffle=True, random_state=42)
    
    best_avg_val_loss = float('inf')
    best_params = {}
    
    # Loop over candidate hyperparameter combinations
    for hu in candidate_hidden_units:
        for reg in candidate_regularizations:
            val_losses = []
            
            # Inner CV loop
            for inner_train_idx, inner_val_idx in inner_cv.split(X_train.cpu().numpy()):
                X_inner_train = X_train[inner_train_idx]
                Y_inner_train = Y_train[inner_train_idx]
                X_inner_val = X_train[inner_val_idx]
                Y_inner_val = Y_train[inner_val_idx]
                
                # Create DataLoader for inner training data
                inner_train_dataset = TensorDataset(X_inner_train, Y_inner_train)
                inner_train_loader = DataLoader(inner_train_dataset, batch_size=32, shuffle=True)
                
                # Define a new model for inner CV with current hyperparameters
                model_inner = torch.nn.Sequential(
                    torch.nn.Linear(X_train.shape[1], hu),
                    torch.nn.LeakyReLU(),
                    torch.nn.Linear(hu, 1)
                ).to(device)
                
                criterion = torch.nn.MSELoss()
                optimizer = torch.optim.Adam(model_inner.parameters(), lr=learning_rate, weight_decay=reg)
                
                # Train the model on the inner training set
                for epoch in range(inner_epochs):
                    for inputs, labels in inner_train_loader:
                        inputs = inputs.to(device)
                        labels = labels.to(device)
                        
                        optimizer.zero_grad()
                        outputs = model_inner(inputs)
                        loss = criterion(outputs, labels.unsqueeze(1))
                        loss.backward()
                        optimizer.step()
                
                # Evaluate on the inner validation set
                model_inner.eval()
                with torch.no_grad():
                    val_outputs = model_inner(X_inner_val)
                    val_loss = criterion(val_outputs, Y_inner_val.unsqueeze(1)).item()
                model_inner.train()
                val_losses.append(val_loss)
            
            # Calculate average validation loss for this hyperparameter combination
            avg_val_loss = sum(val_losses) / len(val_losses)
            if avg_val_loss < best_avg_val_loss:
                best_avg_val_loss = avg_val_loss
                best_params = {'hidden_units': hu, 'regularization': reg}
    
    # Print the best hyperparameters found for this outer fold
    print(f"Fold {fold_idx}: Best params: {best_params} with avg inner val loss: {best_avg_val_loss:.4f}")
    
    # -----------------------------
    # 3b. Retrain final model on the entire outer training set using best hyperparameters
    # -----------------------------
    train_dataset = TensorDataset(X_train, Y_train)
    train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
    
    final_model = torch.nn.Sequential(
        torch.nn.Linear(X_train.shape[1], best_params['hidden_units']),
        torch.nn.LeakyReLU(),
        torch.nn.Linear(best_params['hidden_units'], 1)
    ).to(device)
    
    criterion = torch.nn.MSELoss()
    optimizer = torch.optim.Adam(final_model.parameters(), lr=learning_rate, weight_decay=best_params['regularization'])

    loss_history = []
    
    for epoch in tqdm(range(outer_epochs), desc=f"Fold {fold_idx} Training", leave=False):
        for inputs, labels in train_loader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            
            optimizer.zero_grad()
            outputs = final_model(inputs)
            loss = criterion(outputs, labels.unsqueeze(1))
            loss.backward()
            loss_history.append(loss.item())
            optimizer.step()
    
    # -----------------------------
    # 3c. Evaluate final model on outer test set using the provided MSE formula
    # -----------------------------
    final_model.eval()
    with torch.no_grad():
        test_outputs = final_model(X_test)
        # This computes the MSE as (1/N_test)*sum((y - yhat)^2)
        test_loss = criterion(test_outputs, Y_test.unsqueeze(1)).item()
        final_model.to(device)
        output = final_model(X_tensor).squeeze()
        
        y_plot_data_mapped = map_value(Y_tensor, -1, 1, 0, 4)  # Scale back to 0-4
        output_mapped = map_value(output, -1, 1, 0, 4)   # Scale back to 0-5
        # print(np.round(output_mapped, 2))
        plt.scatter(y_plot_data_mapped.cpu(), output_mapped.cpu(), alpha=0.6)
        
        # Add red dots at the coordinates (0,0), (1,1), (2,2), (3,3), (4,4)
        reference_x = [0, 1, 2, 3, 4]
        reference_y = [0, 1, 2, 3, 4]
        plt.scatter(reference_x, reference_y, color='red', s=50)
        
        plt.xlabel('True Values')
        plt.ylabel('Predicted Values')
        plt.title('Scatter Plot of Predictions vs True Values')
        plt.show()

        plt.figure(figsize=(8, 6))
        plt.plot(loss_history, marker='o', linestyle='-', color='blue')
        plt.title('Loss History')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.grid(True)
        plt.show()
    
    # Baseline model: predict the mean of the outer training targets
    baseline_pred = Y_train.mean()
    baseline_loss = ((Y_test - baseline_pred) ** 2).mean().item()
    
    # Save the results for this fold
    results.append({
        'fold': fold_idx,
        'hidden_units': best_params['hidden_units'],
        'regularization': best_params['regularization'],
        'test_loss': test_loss,
        'baseline_loss': baseline_loss
    })

# -----------------------------
# 4. Create a table with results for each fold
# -----------------------------
results_df = pd.DataFrame(results)
print("\nResults for each fold:")
print(results_df)
```
