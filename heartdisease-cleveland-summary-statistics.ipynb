```python
#male vs female
import scipy.stats as stats

def chi_square_test(observed1, observed2):
    observed = [observed1, observed2]
    
    
    expected = [sum(observed) / 2] * 2
    
    chi2_stat, p_value = stats.chisquare(f_obs=observed, f_exp=expected)
    
    print(f"Chi-i-anden statistik: {chi2_stat:.4f}")
    print(f"P-værdi: {p_value:.4f}")

chi_square_test(97, 206)
```

```python

from scipy.stats import linregress
# Define the data (number of major blood vessels and corresponding proportions)
x = np.array([0, 1, 2, 3])  # Independent variable
y = np.array([0.594059, 0.214521, 0.125413, 0.066007])  # Dependent variable (proportions)

# Perform linear regression
slope, intercept, r_value, p_value, std_err = linregress(x, y)

# Calculate R² value
r_squared = r_value ** 2

# Print the results
print(f"Slope: {slope}")
print(f"Intercept: {intercept}")
print(f"R² Value: {r_squared}")
print(f"P-Value: {p_value}")
print(f"Standard Error: {std_err}")
```
